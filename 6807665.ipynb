```python
import pandas as pd

data = {
    "DeliveryID": [2001, 2002, 2003, 2004, 2004, 2005, 2006],
    "Zone": ["North", "south", "Central", "North", "North", "South", "central"],
    "Restaurant": ["A", "B", "C", "A", "A", "B", "C"],
    "Distance": [5.2, 8.5, None, 4.0, 4.0, 10.5, 6.5],
    "DeliveryFee": [50, 80, 60, 45, 45, None, 65]
}

df = pd.DataFrame(data)
```

```python
# 9.
print(df.isnull().sum())
```

```python
# 10
df = df.drop_duplicates(subset="DeliveryID")
print(df)
```

```python
# 11
df["Zone"] = df["Zone"].str.title()

print(df)
```

```python
# 12
df["Distance"] = df["Distance"].fillna(df["Distance"].median())
```

```python
# 13
df = df.dropna(subset=["DeliveryFee"])

print(df)
```

```python
df["FeePerKm"] = df["DeliveryFee"] / df["Distance"]

print(df)
```

## C. Aggregation

```python
avg_fee_by_zone = df.groupby("Zone")["DeliveryFee"].mean().sort_values(ascending=False)

print(avg_fee_by_zone)
```

```python
avg_fee_by_zone = df.groupby("Zone")["DeliveryFee"].mean()

highest_zone = avg_fee_by_zone.idxmax()
highest_avg_fee = avg_fee_by_zone.max()

print("Zone with the highest average delivery fee:", highest_zone)
print("Highest average delivery fee:", highest_avg_fee)
```

# **Question 2**

```python
sensor_data = [
    "MachineA,72",
    "MachineB,65",
    "MachineA,78",
    "MachineC,81",
    "MachineB,70",
    "MachineA,75",
    "MachineC,76"
]
```

## **A. Mapper**

```python
def mapper(line):
    machine, temperature = line.split(",")
    return (machine, int(temperature))
```

```python
mapper("MachineA,72")
```

## **B. Shuffle / Group**

```python
def shuffle_sort(mapped_pairs):
    grouped = {}

    for machine, temperature in mapped_pairs:
        if machine not in grouped:
            grouped[machine] = []

        grouped[machine].append(temperature)

    return grouped
```

```python
mapped_pairs = []

for line in sensor_data:
    mapped_pairs.append(mapper(line))


grouped_data = shuffle_sort(mapped_pairs)

print(grouped_data)
```

## C. Reducer

```python
def reducer(key, values):
    return (key, max(values))
```

```python
reducer("MachineA", [72, 78, 75])
```

## D. Complete Pipeline

```python
mapped_pairs = []

for line in sensor_data:
    mapped_pairs.append(mapper(line))

print("Mapped Pairs:")
print(mapped_pairs)


grouped_data = shuffle_sort(mapped_pairs)

print("\nGrouped Data:")
print(grouped_data)

final_results = {}

for machine, temperatures in grouped_data.items():
    key, maximum = reducer(machine, temperatures)
    final_results[key] = maximum

print("\nMaximum Temperature by Machine:")
for machine, maximum in final_results.items():
    print(f"{machine}: {maximum}")
```

## **E. Analysis**

```python
highest_machine = max(final_results, key=final_results.get)
highest_temperature = final_results[highest_machine]

print("\nMachine with the Highest Temperature:")
print(f"{highest_machine}: {highest_temperature}")
```

# **Question 3**

```python
usage_data = [
    ("Bangkok", "Video", "User01", 120),
    ("Bangkok", "Music", "User02", 45),
    ("ChiangMai", "Video", "User03", 90),
    ("Bangkok", "Game", "User04", 150),
    ("Phuket", "Music", "User05", 30),
    ("ChiangMai", "Game", "User06", 110),
    ("Phuket", "Video", "User07", 80)
]
```

## A. Create a Spark DataFrame

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("MobileAppUsage") \
    .getOrCreate()


columns = ["city", "service", "user", "usage_minutes"]

df = spark.createDataFrame(usage_data, columns)

df.show()
df.printSchema()
```

## B. Filtering

```python
filtered_df = df.filter(df.usage_minutes >= 90)

filtered_df.show()
```

## C. Aggregation

```python
from pyspark.sql.functions import sum

city_usage = df.groupBy("city") \
    .agg(sum("usage_minutes").alias("total_minutes")) \
    .orderBy("total_minutes", ascending=False)

city_usage.show()
```

## D. Multiple Aggregations

```python
from pyspark.sql.functions import sum, avg, count

service_usage = df.groupBy("service").agg(
    sum("usage_minutes").alias("total_minutes"),
    avg("usage_minutes").alias("avg_minutes"),
    count("*").alias("records")
)

service_usage.show()
```

## E. Distributed Processing Concept

```python
rdd = spark.sparkContext.parallelize(usage_data, 3)

print("Number of partitions:", rdd.getNumPartitions())

print("Partition contents:")
print(rdd.glom().collect())
```
