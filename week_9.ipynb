## **Step 2 — Create a SparkSession**

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("CSC475Week9Lab")
    .getOrCreate()
)

print("Spark version:", spark.version)
```

## **Step 3 — Create the Sales DataFrame**

```python
sales_data = [
    (1001, "Bangkok",   "Electronics", "Laptop",   1, 35000.0),
    (1002, "Bangkok",   "Accessories", "Mouse",    2,   500.0),
    (1003, "ChiangMai", "Electronics", "Laptop",   1, 34000.0),
    (1004, "Bangkok",   "Accessories", "Keyboard", 1,  1500.0),
    (1005, "Phuket",    "Food",        "Coffee",   3,   120.0),
    (1006, "ChiangMai", "Food",        "Coffee",   2,   100.0),
    (1007, "Phuket",    "Electronics", "Monitor",  1,  9000.0),
    (1008, "Bangkok",   "Electronics", "Monitor",  2,  8500.0),
    (1009, "ChiangMai", "Accessories", "Mouse",    3,   450.0),
    (1010, "Phuket",    "Accessories", "Keyboard", 2,  1400.0),
]

columns = [
    "order_id", "city", "category",
    "product", "quantity", "unit_price"
]

df = spark.createDataFrame(sales_data, columns)
df.show()
```

## **Step 4 — Inspect the DataFrame**

```python
df.printSchema()
print("Columns:", df.columns)
print("Number of rows:", df.count())
df.describe().show()
```

## **Step 5 — Select Columns**

```python
selected_df = df.select(
    "order_id",
    "city",
    "product",
    "unit_price"
)

selected_df.show()
```

## **Step 6 — Use Column Expressions**

```python
from pyspark.sql.functions import col

price_preview = df.select(
    "product",
    col("unit_price"),
    (col("unit_price") * 1.07).alias("price_with_tax")
)

price_preview.show()
```

## **Step 7 — Filter Rows with One Condition**

```python
high_price_df = df.filter(
    col("unit_price") > 1000
)

high_price_df.show()
```

## **Step 8 — Filter Rows with Multiple Conditions**

```python
bangkok_high_price = df.filter(
    (col("city") == "Bangkok") &
    (col("unit_price") > 1000)
)

bangkok_high_price.show()
```

## **Step 9 — Use where() as an Alternative to filter()**

```python
electronics_df = df.where(
    col("category") == "Electronics"
)

electronics_df.show()
```

## **Step 10 — Predict a Filter Result**

```python
accessory_multi_qty = df.filter(
    (col("category") == "Accessories") &
    (col("quantity") >= 2)
)

accessory_multi_qty.show()
```

## **Step 11 — Create a Derived Column: total_value**

```python
df2 = df.withColumn(
    "total_value",
    col("quantity") * col("unit_price")
)

df2.show()
```

## **Step 12 — Chain Multiple Transformations**

```python
df3 = (
    df2
    .withColumn(
        "tax",
        col("total_value") * 0.07
    )
    .withColumn(
        "final_value",
        col("total_value") * 1.07
    )
)

df3.select(
    "order_id", "total_value", "tax", "final_value"
).show()
```

## **Step 13 — Rename and Drop Columns**

```python
renamed_df = df3.withColumnRenamed(
    "unit_price",
    "price"
)

reduced_df = renamed_df.drop("tax")

reduced_df.printSchema()
```

## **Step 14 — Cast a Column to Another Data Type**

```python
df_cast = df.withColumn(
    "quantity",
    col("quantity").cast("double")
)

df_cast.printSchema()
```

## **Step 15 — Basic Missing-Data Handling**

```python
sales_with_missing = sales_data + [
    (1011, "Bangkok", "Food", "Tea", None, 80.0),
    (1012, "Phuket", "Accessories", "Cable", 2, None),
]

missing_df = spark.createDataFrame(
    sales_with_missing,
    columns
)

missing_df.show()

# Option 1: remove rows containing any null value
missing_df.na.drop().show()

# Option 2: fill selected numeric nulls
filled_df = missing_df.na.fill({
    "quantity": 0,
    "unit_price": 0.0
})

filled_df.show()
```

## **Step 16 — Sort Transactions by total_value**

```python
top_orders = df2.orderBy(
    col("total_value").desc()
)

top_orders.select(
    "order_id", "city", "product", "total_value"
).show()
```

## **Step 17 — Aggregate Total Sales by City**

```python
city_sales = (
    df2
    .groupBy("city")
    .sum("total_value")
)

city_sales.show()
```

## **Step 18 — Use agg() for Multiple Statistics**

```python
from pyspark.sql import functions as F

city_summary = (
    df2
    .groupBy("city")
    .agg(
        F.sum("total_value").alias("total_sales"),
        F.avg("total_value").alias("avg_order_value"),
        F.count("*").alias("transactions")
    )
    .orderBy(col("total_sales").desc())
)

city_summary.show()
```

## **Step 19 — Analyze Product Categories**

```python
category_summary = (
    df2
    .groupBy("category")
    .agg(
        F.sum("total_value").alias("total_sales"),
        F.avg("total_value").alias("avg_sales"),
        F.max("total_value").alias("max_order"),
        F.count("*").alias("transactions")
    )
    .orderBy(col("total_sales").desc())
)

category_summary.show()
```

## **Step 20 — Filter an Aggregated Result**

```python
category_summary.filter(
    col("total_sales") > 5000
).show()
```

## **Step 21 — Create a Temporary View for Spark SQL**

```python
df2.createOrReplaceTempView("sales")

spark.sql("""
SELECT *
FROM sales
""").show()
```

## **Step 22 — Filter and Sort with Spark SQL**

```python
spark.sql("""
SELECT
    order_id,
    city,
    product,
    total_value
    FROM sales
WHERE total_value > 5000
ORDER BY total_value DESC
""").show()
```

## **Step 23 — Aggregate with Spark SQL**

```python
spark.sql("""
SELECT
    city,
    SUM(total_value) AS total_sales,
    AVG(total_value) AS avg_order_value,
    COUNT(*) AS transactions
FROM sales
GROUP BY city
ORDER BY total_sales DESC
""").show()
```

## **Step 24 — Solve the Same Problem with Two APIs**

```python
# DataFrame API
api_result = (
    df2
    .groupBy("city")
    .agg(F.sum("total_value").alias("total_sales"))
    .orderBy(col("total_sales").desc())
)

api_result.show()

# Spark SQL
sql_result = spark.sql("""
SELECT city, SUM(total_value) AS total_sales
FROM sales
GROUP BY city
ORDER BY total_sales DESC
""")

sql_result.show()
```

## **Step 26 — Repartition the DataFrame**

```python
df_repartitioned = df2.repartition(4)

print(
    "Partitions after repartition:",
    df_repartitioned.rdd.getNumPartitions()
)

print(
    "Rows per partition:",
    df_repartitioned.rdd.glom().map(len).collect()
)
```

## **Step 27 — Inspect an Execution Plan with explain()**

```python
plan_df = (
    df2
    .filter(col("total_value") > 1000)
    .groupBy("city")
    .sum("total_value")
)

plan_df.explain()
```

## **Step 28 — Observe Lazy Evaluation**

```python
pipeline = (
    df2
    .filter(col("total_value") > 1000)
    .select("city", "category", "total_value")
    .groupBy("city")
    .agg(
        F.sum("total_value").alias("total_sales")
    )
)

# The pipeline has been defined.
# Requesting output triggers execution.
pipeline.show()
```

## **Step 29 — Build a Complete Analytics Pipeline**

```python
result = (
    df
    .withColumn(
        "total_value",
        col("quantity") * col("unit_price")
    )
    .filter(
        col("total_value") > 1000
    )
    .groupBy("city")
    .agg(
        F.sum("total_value").alias("total_sales"),
        F.avg("total_value").alias("avg_order_value")
    )
    .orderBy(
        col("total_sales").desc()
    )
)

result.show()
```

## **Step 30 — Create an Order-Level Classification**

```python
from pyspark.sql.functions import when

classified_df = df2.withColumn(
    "order_level",
    when(col("total_value") >= 20000, "High")
    .when(col("total_value") >= 2000, "Medium")
    .otherwise("Low")
)

classified_df.select(
    "order_id", "total_value", "order_level"
).show()
```
